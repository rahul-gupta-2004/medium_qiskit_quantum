Importing the Libraries

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import numpy as np
```

Building the Oracle

```python
# Constant Oracle (return 0)

def constant_oracle():
    qc = QuantumCircuit(2)
    return qc
```

```python
# Constant Oracle (returns 1)

def constant_oracle_one():
    qc = QuantumCircuit(2)
    qc.x(1)
    return qc
```

```python
# Balanced Oracle

def balanced_oracle():
    qc = QuantumCircuit(2)
    qc.cx(0, 1)
    return qc
```

The Deutsch‑Jozsa Circuit

```python
def deutsch_jozsa_circuit(oracle):
    qc = QuantumCircuit(2, 1)
    
    qc.x(1)
    qc.h(0)
    qc.h(1)
    
    qc.compose(oracle, inplace=True)
    
    qc.h(0)
    qc.measure(0, 0)
    
    return qc
```

Running the Simulation

Testing the Constant Oracle (returns 0)

```python
const_oracle = constant_oracle()
circuit_const = deutsch_jozsa_circuit(const_oracle)

simulator = AerSimulator()
job = simulator.run(circuit_const, shots=1024)
counts = job.result().get_counts()

print("Constant Oracle (returns 0):", counts)
```

Testing the Balanced Oracle

```python
bal_oracle = balanced_oracle()
circuit_bal = deutsch_jozsa_circuit(bal_oracle)

job = simulator.run(circuit_bal, shots=1024)
counts = job.result().get_counts()

print("Balanced Oracle:", counts)
```

Visualizing the Circuits

```python
print("Constant Oracle Circuit:")
print(const_oracle.draw())
print("\nBalanced Oracle Circuit:")
print(bal_oracle.draw())
print("\nFull Deutsch-Jozsa Circuit with Balanced Oracle:")
print(circuit_bal.draw())
```
