Import the libraries

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
```

Build a circuit with conditional logic

```python
qc = QuantumCircuit(2, 1)

qc.h(0)
qc.measure(0, 0)
with qc.if_test((0, 1)):
    qc.x(1)
qc.measure(1, 0)
```

Simulate and collect results

```python
sim = AerSimulator()
transpiled = transpile(qc, sim)
result = sim.run(transpiled, shots=1024).result()
counts = result.get_counts()

print(counts)
```

Verify the conditional effect

```python
qc_no_cond = QuantumCircuit(2, 1)
qc_no_cond.h(0)
qc_no_cond.measure(0, 0)
qc_no_cond.measure(1, 0)    # direct measurement, no X

transpiled_no = transpile(qc_no_cond, sim)
result_no = sim.run(transpiled_no, shots=1024).result()
counts_no = result_no.get_counts()

print(counts_no)
```
